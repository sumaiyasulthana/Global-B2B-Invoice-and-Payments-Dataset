Dropping Unwanted columns for better feature Engineering Steps

```python
columns_to_drop = [
    'invoice_id', 'invoice_number', 'customer_email',
    'customer_name', 'vendor_name', 'product_name',
    'subtotal', 'discount_amount', 'tax_amount', 'payment_date'
]
a = df1.drop(columns=columns_to_drop)
a.head()
```

```python
b = df2.drop(columns=columns_to_drop)
b.head()
```

```python
df=pd.concat([a,b],axis=0)
df.head()
```

```python
df.shape
```

```python
df.info()
```

Treat Missing Values

```python
df.isnull().sum()
```

```python
df['customer_city'].unique()
```

filling values by mean for numerics

```python

df['quantity'].fillna(df['quantity'].mean(),inplace=True)
df['unit_price'].fillna(df['unit_price'].mean(),inplace=True)
df['discount_percent'].fillna(df['discount_percent'].mean(),inplace=True)
df['tax_percent'].fillna(df['tax_percent'].mean(),inplace=True)
df['shipping_cost'].fillna(df['shipping_cost'].mean(),inplace=True)
df['total_amount'].fillna(df['total_amount'].mean(),inplace=True)

```

filling values by mode for categorical Data

```python
df['customer_city'].fillna(df['customer_city'].mode()[0],inplace=True)
df['customer_state'].fillna(df['customer_state'].mode()[0],inplace=True)
df['customer_country'].fillna(df['customer_country'].mode()[0],inplace=True)
df['product_category'].fillna(df['product_category'].mode()[0],inplace=True)
df['currency'].fillna(df['currency'].mode()[0],inplace=True)
df['payment_method'].fillna(df['payment_method'].mode()[0],inplace=True)
df['payment_status'].fillna(df['payment_status'].mode()[0],inplace=True)
df['order_channel'].fillna(df['order_channel'].mode()[0],inplace=True)
df['sales_rep'].fillna(df['sales_rep'].mode()[0],inplace=True)
```

filled wih 0 by seeing it has nly 2 unique count and most appeared is 0

```python
df['is_overdue'].fillna(0,inplace=True)
```

```python
id = ['customer_id', 'vendor_id', 'product_id']
df.dropna(subset=id)
```
